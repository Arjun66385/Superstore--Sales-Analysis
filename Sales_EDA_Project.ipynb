```python
# Load the dataset
df = pd.read_csv("/content/Sample - Superstore.csv", encoding="latin1")

# Display dataset information
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display first 5 rows
df.head()
```

```python
# Dataset information
df.info()

# Statistical summary
df.describe()

# Column names
print("Columns:", df.columns.tolist())
```

```python
# Convert date columns
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Verify data types
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
# Calculate key business metrics
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_quantity = df["Quantity"].sum()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Quantity Sold:", total_quantity)
```

```python
# Calculate sales by category
category_sales = df.groupby("Category")["Sales"].sum().sort_values()

# Create bar chart
category_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum().sort_values()

plt.figure(figsize=(8, 5))
region_profit.plot(kind="bar")

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = (
    df.set_index("Order Date")
      .resample("ME")["Sales"]
      .sum()
)

plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .nlargest(10)
      .sort_values()
)

plt.figure(figsize=(10, 6))
top_products.plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum()

plt.figure(figsize=(7, 5))
segment_sales.plot(kind="pie", autopct="%1.1f%%", startangle=90)

plt.title("Sales Distribution by Customer Segment")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Discount", y="Profit", alpha=0.5)

plt.title("Relationship Between Discount and Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

```python
print("===== SALES ANALYSIS SUMMARY =====")

print("\n1. Total Sales:", round(df["Sales"].sum(), 2))
print("2. Total Profit:", round(df["Profit"].sum(), 2))
print("3. Total Orders:", df["Order ID"].nunique())
print("4. Total Quantity Sold:", df["Quantity"].sum())

print("\n5. Best Category by Sales:")
print(df.groupby("Category")["Sales"].sum().idxmax())

print("\n6. Most Profitable Region:")
print(df.groupby("Region")["Profit"].sum().idxmax())

print("\n7. Most Profitable Category:")
print(df.groupby("Category")["Profit"].sum().idxmax())

print("\n8. Top 5 States by Sales:")
print(df.groupby("State")["Sales"].sum().nlargest(5))
```

```python
# Sales and profit by category
category_summary = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).round(2)

print("SALES AND PROFIT BY CATEGORY")
display(category_summary)

# Sales and profit by region
region_summary = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).round(2)

print("SALES AND PROFIT BY REGION")
display(region_summary)
```
